## 3. Background

### 3.1 What Is the Jetson Orin Nano Super?
A small computer built for AI at the edge. It looks like a hobby board, but it carries a real NVIDIA GPU, so it can run camera and AI workloads that would overwhelm a typical single-board computer. "Super" refers to Super Mode, which JetPack 6.2 unlocks. It raises the clock speeds and the memory bandwidth of the same hardware.

| Item | Specification (8 GB module, Super Mode) |
|---|---|
| AI performance | Up to 67 TOPS (INT8, sparse) |
| GPU | NVIDIA Ampere, 1024 CUDA cores, 32 Tensor cores |
| CPU | 6-core Arm Cortex-A78AE |
| Memory | 8 GB LPDDR5, **shared** between CPU and GPU, about 102 GB/s |
| Storage | M.2 Key M slot for an NVMe SSD (preinstalled) |
| Network | Gigabit Ethernet and an M.2 Key E Wi-Fi module |
| Other I/O | USB Type-A ports, USB-C, 2 camera (CSI) connectors, 40-pin header |

> **Note.** The memory is shared between the CPU and the GPU. A program that uses 3 GB of RAM leaves less for the GPU. You will see this in jtop later.

### 3.2 What Is JetPack?
A new Jetson has no operating system. JetPack is NVIDIA's software bundle that gives it one, with the libraries needed for GPU computing and AI. JetPack 6.2 is built on Jetson Linux 36.4.3 (Linux kernel 5.15, Ubuntu 22.04).

| Component | What it does |
|---|---|
| Jetson Linux (L4T) | Bootloader, Linux kernel, drivers and the Ubuntu 22.04 file system |
| CUDA | Lets programs run general computing on the GPU (Lab 5) |
| cuDNN | GPU building blocks for neural networks |
| TensorRT | Makes trained models run faster on the Jetson (Lab 8) |
| Other libraries | Multimedia, computer vision and camera tools such as VPI and DeepStream |

You confirm the real version numbers on your own board in Part D, so you do not need to memorise this table.

### 3.3 Power Modes
A lower power limit saves energy and makes less heat. A higher limit lets the CPU and GPU run faster.

| Mode | Typical use | Trade-off |
|---|---|---|
| 15W | Battery or small enclosures | Lowest heat and power, lowest speed |
| 25W | Default on the developer kit | Good balance for most work |
| MAXN SUPER | Maximum performance | Highest clocks, so more heat and a louder fan |

### Compare Power Modes
For each mode this cell switches, waits 30 seconds for the clocks to settle, then prints three seconds of `tegrastats` (CPU load, memory, GPU load, temperatures and power). Watch the same numbers on the CTRL and ALL pages of jtop in the other tab.

```python
%%bash
for m in 1 2; do
  sudo nvpmodel -m $m
  sudo nvpmodel -q | head -n 1
  sleep 30
  sudo timeout 3 tegrastats || true     # timeout stops tegrastats on purpose; that is not an error
done
```

## 7. Troubleshooting

| Symptom | Likely cause | What to try |
|---|---|---|
| No picture on the monitor | The monitor was not ready, the wrong cable end or input | Check the monitor power and input. Switch the Jetson off and on again with the monitor already on. |
| The board does not start | The power supply or the power switch | Check the 19 V plug and the power switch on the case. Try another socket. |
| Keyboard or mouse does not work | Receiver not plugged in, or flat batteries | Check the receiver in a USB Type-A port. Replace the batteries. |
| The Wi-Fi network is not listed | Antennas loose, or out of range | Check the antennas. Move closer or use Ethernet. |
| jtop says the service is not running | The board was not restarted after jetson-stats was installed | Run the `systemctl restart jtop.service` cell, or `sudo reboot`. |
| `sudo` rejects your password | Typing error. Nothing shows while you type. | Run the `sudo_login` cell again and type slowly. |
| The first boot is very slow | The system prepares itself the first time | Wait several minutes. Do not switch the board off. |
| Fan runs loudly | Normal in MAXN SUPER | Switch to 25W if the noise is a problem. |
