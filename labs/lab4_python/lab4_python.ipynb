## 2. Background

**Two ways to run Python.** The *interactive prompt* (type `python3`, then lines after `>>>`) is best for trying one idea. A *script file* (write a `.py` file and run `python3 file.py`) is best for programs you keep. A notebook cell, like the ones below, works like the prompt but keeps your code and results on one page.

**Virtual environments and pip.** Different projects need different versions of libraries. A virtual environment is a private folder with its own Python packages, so one project cannot break another or the system. `pip` installs packages into it. On the Jetson the course environment is made with `--system-site-packages`, so it can also see JetPack libraries installed with apt, such as the TensorRT bindings and OpenCV.

| Command | What it does |
|---|---|
| `python3 -m venv --system-site-packages DIR` | Creates an environment in DIR |
| `source DIR/bin/activate` | Turns it on. The prompt starts with the environment name. |
| `deactivate` | Turns it off |
| `pip install name` | Installs a package |
| `pip list` | Shows installed packages |
| `pip freeze --local > requirements.txt` | Saves the package list to a file |

> **Warning.** Never use `sudo` with pip inside your environment. It creates files owned by root that you cannot change later. A plain `pip install torch` often gives a version without GPU support on the Jetson. The correct build was installed in Lab 0, Part 0.

**Python building blocks.**

| Idea | Example | Notes |
|---|---|---|
| Variable | `cores = 6` | A name that holds a value |
| Types | `6`, `1.5`, `"text"`, `True` | Whole number, decimal, text, yes/no |
| List | `modes = ["15W", "25W"]` | Ordered. The first item is number 0. |
| Dictionary | `limits = {"15W": 15}` | Pairs of key and value |
| f-string | `f"{name} has {cores} cores"` | Puts values inside text |
| Condition | `if x > 5:` then `elif`, `else` | Chooses what to do |
| Loops | `for item in items:` and `while x < 3:` | Repeat code |
| Function | `def square(n):` | A named block of code you can reuse |
| Import | `import math` | Loads a library |

> **Note.** Indentation is part of Python. The lines inside an `if`, a `for` or a `def` must be indented by the same amount. Use 4 spaces and never mix in tabs. A missing colon `:` at the end of those lines is the most common beginner error.

**NumPy and arrays.** NumPy provides the *array*, a grid of numbers of one type. An array has a *shape* (its size in each direction) and a *dtype* (the type of number). NumPy does maths on the whole array at once, in fast compiled code, so you rarely need a loop. Images are arrays too. Every pixel is a number from 0 (black) to 255 (white) stored as `uint8`.

| Image | Shape | Example |
|---|---|---|
| Grayscale | (height, width) | (480, 640) |
| Colour | (height, width, 3) | (480, 640, 3) |

### Script 1: Variables and Maths

````python
board = "Jetson Orin Nano Super"
cpu_cores = 6
memory_gb = 8
power_modes = ["15W", "25W", "MAXN SUPER"]

print("Board:", board)
print(f"{board} has {cpu_cores} CPU cores and {memory_gb} GB of memory")
print("Memory per core:", memory_gb / cpu_cores)
print("Whole GB per core:", memory_gb // cpu_cores)
print("Left over:", memory_gb % cpu_cores)
print("Type of cpu_cores:", type(cpu_cores))
print("Number of power modes:", len(power_modes))
````

### Script 2: Lists, Conditions and Loops

````python
power_modes = ["15W", "25W", "MAXN SUPER"]
limits = {"15W": 15, "25W": 25, "MAXN SUPER": None}

for mode in power_modes:
    limit = limits[mode]
    if limit is None:
        print(mode, "has no fixed power limit")
    elif limit >= 25:
        print(mode, "is a high power mode")
    else:
        print(mode, "is a low power mode")

count = 0
while count < 3:
    print("Loop number", count)
    count += 1

squares = [n * n for n in range(1, 6)]
print("Squares:", squares)
````

Expected output:
```
15W is a low power mode
25W is a high power mode
MAXN SUPER has no fixed power limit
Loop number 0
Loop number 1
Loop number 2
Squares: [1, 4, 9, 16, 25]
```
### Script 3: Functions, Files and Errors

````python
import math


def celsius_to_fahrenheit(c):
    return c * 9 / 5 + 32


def describe_temperature(c):
    if c < 50:
        return "cool"
    elif c < 70:
        return "warm"
    return "hot"


for t in [35, 48.5, 62, 85]:
    f = celsius_to_fahrenheit(t)
    print(f"{t} C = {f:.1f} F ({describe_temperature(t)})")

print("Square root of 2:", math.sqrt(2))

with open("readings.txt", "w") as file:
    for t in [35, 48.5, 62, 85]:
        file.write(f"{t}\n")

with open("readings.txt") as file:
    values = [float(line) for line in file]
print("Read back:", values)
print("Average:", sum(values) / len(values))

try:
    number = int("abc")
except ValueError:
    print("That is not a number")
````

### Step 3: Compare a Python Loop with NumPy
Run the cell **three times**. Your numbers differ from run to run and from your neighbour's. NumPy is usually 10 to 100 times faster.

````python
import time
import numpy as np

n = 1_000_000
data = list(range(n))

start = time.perf_counter()
total = 0
for x in data:
    total += x * 2
loop_time = time.perf_counter() - start

arr = np.arange(n)
start = time.perf_counter()
total_np = (arr * 2).sum()
numpy_time = time.perf_counter() - start

print(f"Python loop : {loop_time * 1000:.1f} ms")
print(f"NumPy       : {numpy_time * 1000:.1f} ms")
print(f"Speed-up    : {loop_time / numpy_time:.0f}x")
````
