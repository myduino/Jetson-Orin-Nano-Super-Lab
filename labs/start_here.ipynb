## How to Use a Lab Notebook
- Run a cell with **Shift + Enter**. Read the code before you run it.
- A cell that starts with `!` runs **one Linux command**. A cell that starts with `%%bash` is a small **Linux session**: every line is a command, run in order. A cell that starts with `%pip` installs a Python package. The output appears under the cell, while it runs.
- Commands with `sudo` need your password. A notebook has no terminal to ask for it, so a cell runs `labview.sudo_login()` once. Type your password in the box that appears (nothing is shown).
- Programs that fill the whole screen and wait for keys (`jtop`, `nano`, `less`, `htop`) need a real terminal. Open one **inside JupyterLab**: **File, New, Terminal**. Drag its tab to the side of the screen to see it beside the notebook. It is already in your own environment.
- Answer questions in the cells that say **Your answers**. Double-click the cell, type, then press Shift + Enter.
- Press **Ctrl + S** often. Your copy is saved on the Jetson.
- If a cell fails, run the cells above it again from the top. **Kernel, Restart Kernel** helps when it behaves strangely. The square **stop** button stops a cell that never finishes.
- Show your instructor each Check Task at the end of the lab.

## Live Camera in the Notebook
The labs that use the camera show the live picture **under the cell**, with a **Stop** button. Press **Stop** before you run another cell that uses the camera. If you see `Cannot open the camera`, tell your teacher: the camera is chosen on the teacher page.

## Where Your Work Is Saved
Everything you save is in **your own folder**, `~/students/yourname`, one folder for each lab (`lab4`, `lab5` and so on). Other students of this Jetson have their own folders and cannot change yours by accident. The lab code and the original notebooks stay in `~/Jetson-Orin-Nano-Super-Lab`, so updating that folder never changes your answers.

## When You Finish
Before you leave, copy your folder to your laptop or a USB drive. Your teacher deletes the folders of the session after the lab.

## If You Break a Notebook
Open a Terminal tab inside JupyterLab (**File, New, Terminal**) and type `lab reset 4` (use your lab number). You get a fresh copy. Your old copy is kept next to it with `.old` at the end of the name.

## Safety
- Shut down with `sudo shutdown now`. Do not pull the power cable.
- The heatsink and case get hot in MAXN SUPER mode. Do not touch them.
- Do not run `sudo apt upgrade` or `pip install torch` unless your instructor says so.
- The camera sees faces. Do not share the pictures that the labs save.
