# Jetson Orin Nano Super Lab: Start Here

Welcome. Every lab of this course is a notebook like this one. You read, run the commands and write your answers in the same place. You do not switch between a PDF, a terminal and the screen: the Linux commands, the Python code and the live camera all run **inside the notebook**.

## Every time you start a lab
1. On your laptop open a terminal (on Windows: PowerShell) and connect to the Jetson. Your teacher gives you the address:
   `ssh orin_nano@<address of the Jetson>`
2. Start your own environment. Use your own name, always the same, and the camera number your teacher gives you:
   `lab start yourname class yourclass camera 0`
   (Leave out `camera 0` if your lab does not use a camera. `lab cameras` shows which cameras are free.)
3. It prints a **JupyterLab link**. Copy it into the browser of your laptop. This page opens.
4. Click the lab you want in the table below. From now on you work only in the browser.

Your JupyterLab always has **the same port number** (the number after the `:` in the link), so you can bookmark it. It is shown on the teacher's page too.

Keep the SSH terminal open in the background: it keeps your JupyterLab running.

## When you finish
Save your notebooks (**Ctrl + S**), copy your folder to your laptop, then type `exit` in the SSH terminal. That stops your JupyterLab and frees your camera. Do not pull the power cable of the Jetson.

(At the Jetson's own keyboard and monitor, `lab start yourname class yourclass` works too and opens the browser on that screen.)

## How to Use a Lab Notebook
- Run a cell with **Shift + Enter**. Read the code before you run it.
- A cell that starts with `!` runs **one Linux command**. A cell that starts with `%%bash` is a small **Linux session**: every line is a command, run in order. A cell that starts with `%pip` installs a Python package. The output appears under the cell, while it runs.
- Commands with `sudo` need your password. A notebook has no terminal to ask for it, so a cell runs `labview.sudo_login()` once. Type your password in the box that appears (nothing is shown).
- Programs that fill the whole screen and wait for keys (`jtop`, `nano`, `less`, `htop`) need a real terminal. Open one **inside JupyterLab**: **File, New, Terminal**. Drag its tab to the side of the screen to see it beside the notebook. It is already in your own environment.
- Answer questions in the cells that say **Your answers**. Double-click the cell, type, then press Shift + Enter.
- Press **Ctrl + S** often. Your copy is saved on the Jetson.
- If a cell fails, run the cells above it again from the top. **Kernel, Restart Kernel** helps when it behaves strangely. The square **stop** button stops a cell that never finishes.
- Show your instructor each Check Task at the end of the lab.

## Live Camera in the Notebook
The labs that use the camera show the live picture **under the cell**, with a **Stop** button. Press **Stop** before you run another cell that uses the camera. If you see `Cannot open the camera`, start again with `lab start yourname class yourclass camera 0`.

## Where Your Work Is Saved
Everything you save is in **your own folder**, `~/myduino-jetson-lab/yourclass/yourname`, one folder for each lab (`lab4`, `lab5` and so on). Other students of this Jetson have their own folders and cannot change yours by accident. The lab code and the original notebooks stay in `~/Jetson-Orin-Nano-Super-Lab`, so updating that folder never changes your answers.

## When You Finish
Before you leave, copy your folder to your laptop or a USB drive. Your teacher deletes the folders of the session after the lab.

## If You Break a Notebook
Open a Terminal tab (**File, New, Terminal**) and type `lab reset 4` (use your lab number). You get a fresh copy. Your old copy is kept next to it with `.old` at the end of the name.

## Safety
- Shut down with `sudo shutdown now`. Do not pull the power cable.
- The heatsink and case get hot in MAXN SUPER mode. Do not touch them.
- Do not run `sudo apt upgrade` or `pip install torch` unless your instructor says so.
- The camera sees faces. Do not share the pictures that the labs save.
