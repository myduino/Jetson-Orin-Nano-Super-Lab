# Jetson Orin Nano Super Lab: Start Here

Welcome. Every lab of this course is a notebook like this one. You read, run the code and write your answers in the same place, so you do not switch between a PDF, a terminal and the screen.

## Every time you start a lab
1. On your laptop open a terminal (on Windows: PowerShell) and connect to the Jetson. Your teacher gives you the address:
   `ssh orin_nano@<address of the Jetson>`
2. Start your own environment. Use your own name, always the same, and the camera number your teacher gives you:
   `lab start yourname camera 0`
   (Leave out `camera 0` if your lab does not use a camera. `lab cameras` shows which cameras are free.)
3. It prints a **JupyterLab link**. Copy it into the browser of your laptop. This page opens.
4. Click the lab you want in the table below.

In the terminal you are now in your own environment: your folder, your Python and your camera are ready, and the prompt starts with your name. Programs that show live video print a second link. Open it in the browser of your laptop and press the keys in the terminal.

## When you finish
Save your notebooks (**Ctrl + S**), copy your folder to your laptop, then type `exit` in the SSH terminal. That stops your JupyterLab and frees your camera. Do not pull the power cable of the Jetson.

(At the Jetson's own keyboard and monitor, `lab start yourname` works too and opens the browser on that screen.)

## How to Use a Lab Notebook
- Run a cell with **Shift + Enter**. Read the code before you run it.
- Answer questions in the cells that say **Your answers**. Double-click the cell, type, then press Shift + Enter.
- Some steps need the terminal. Open one inside JupyterLab: **File, New, Terminal**. Drag its tab to the side of the screen to see it beside the notebook.
- Press **Ctrl + S** often. Your copy is saved on the Jetson.
- If a cell fails, run the cells above it again from the top. **Kernel, Restart Kernel** helps when it behaves strangely.
- Show your instructor each Check Task at the end of the lab.

## Where Your Work Is Saved
Everything you save is in **your own folder**, `~/students/yourname`, one folder for each lab (`lab4`, `lab5` and so on). Other students of this Jetson have their own folders and cannot change yours by accident. The lab code and the original notebooks stay in `~/Jetson-Orin-Nano-Super-Lab`, so updating that folder never changes your answers.

## When You Finish
Before you leave, copy your folder to your laptop or a USB drive. Your teacher deletes the folders of the session after the lab.

## If You Break a Notebook
Close it, then type `lab reset 4` (use your lab number) in the terminal. You get a fresh copy. Your old copy is kept next to it with `.old` at the end of the name.

## Safety
- Shut down with `sudo shutdown now`. Do not pull the power cable.
- The heatsink and case get hot in MAXN SUPER mode. Do not touch them.
- Do not run `sudo apt upgrade` or `pip install torch` unless your instructor says so.
- The camera sees faces. Do not share the pictures that the labs save.
