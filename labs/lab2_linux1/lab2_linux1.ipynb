## 3. Background

### 3.1 What Are Linux and Ubuntu?
Linux is the core of the operating system, called the kernel. It talks to the hardware. Ubuntu is a complete package built around that kernel: the kernel, thousands of tools, a software installer and a desktop. JetPack 6.2 uses Ubuntu 22.04, the same Ubuntu found on many servers and PCs. The only difference is that it runs on the Jetson's Arm processor.

### 3.2 The Terminal and the Shell
The terminal is the window. The shell (here, Bash) is the program inside it that reads what you type and runs it. When the shell is ready it shows a prompt:

`orin_nano@jetson-01:~$`

| Part | Meaning |
|---|---|
| `orin_nano` | Your username |
| `@` | Means "at" |
| `jetson-01` | The computer name |
| `:` | A separator |
| `~` | The folder you are in now. `~` means your home folder. |
| `$` | The shell is ready for a command. A `#` here would mean you are the administrator. |

### 3.3 How a Command Is Written
Most commands follow the pattern: the command, then options, then arguments. Parts are separated by spaces.

`ls -l /home`

| Part | Example | Meaning |
|---|---|---|
| Command | `ls` | What to do (list the contents of a folder) |
| Option | `-l` | How to do it (long format with more details) |
| Argument | `/home` | What to act on |

> **Note.** Linux is case sensitive. `Notes.txt`, `notes.txt` and `NOTES.TXT` are three different files. Avoid spaces in file names: use `my_notes.txt`.

### 3.4 The File System Tree
Linux has no C: or D: drive. Everything lives in one tree that starts at a single point called the root, written `/`. Your own files live in your home folder, `/home/` followed by your username.

```
/                      (root: the top of everything)
├── home/
│   └── orin_nano/     (your home folder, shown as ~)
├── etc/               (system settings)
├── usr/               (installed programs and libraries)
├── var/               (logs and data that change)
├── dev/               (devices appear as files)
├── tmp/               (temporary files)
└── opt/               (optional add-on software)
```

### 3.5 Absolute and Relative Paths
A path is the address of a file or folder. An **absolute** path starts with `/` and works from anywhere, for example `/home/orin_nano/Documents`. A **relative** path starts from the folder you are in now.

| Symbol | Meaning |
|---|---|
| `/` | The root folder (or the separator between folder names) |
| `~` | Your home folder |
| `.` | The current folder |
| `..` | The parent folder, one level up |
| `-` | Used with `cd`: the folder you were in before |

### 3.6 Command Quick Reference

| Command | What it does | Example |
|---|---|---|
| `pwd` | Shows the folder you are in | `pwd` |
| `ls` | Lists the contents of a folder | `ls -l` |
| `cd` | Changes folder | `cd Documents` |
| `mkdir` | Makes a new folder | `mkdir data` |
| `touch` | Creates an empty file | `touch a.txt` |
| `cp` | Copies a file | `cp a.txt b.txt` |
| `mv` | Moves or renames a file | `mv a.txt notes/` |
| `rm` | Deletes a file (permanent) | `rm b.txt` |
| `rmdir` | Deletes an empty folder | `rmdir data` |
| `cat` | Prints a whole file | `cat a.txt` |
| `head` | Prints the first lines of a file | `head -n 3 a.txt` |
| `less` | Pages through a file (`q` to quit) | `less /etc/nvpmodel.conf` |
| `echo` | Prints text | `echo hello` |
| `nano` | Edits a text file | `nano a.txt` |
| `man` | Opens the manual for a command | `man ls` |
| `clear` | Clears the screen | `clear` |

### Step 1: List Files in Different Ways
Start in your home folder and try each form of `ls`. Compare the output.

````python
%%bash
cd ~
ls
echo "---- ls -l"
ls -l
echo "---- ls -a"
ls -a
echo "---- ls -lh"
ls -lh
````

- `-l` shows the long format: permissions, owner, size and date.
- `-a` shows all files, including hidden ones. Hidden files start with a dot, such as `.bashrc`.
- `-h` with `-l` shows sizes in readable units such as K and M.

````python
%%bash
cd /
ls
cd /etc
ls
cd ~
pwd
cd /usr
cd ..
pwd
cd -
````

Typing `cd` with nothing after it also takes you home. `cd ..` goes up one level and `cd -` jumps back to the previous folder.

## 7. Troubleshooting

| Symptom | Likely cause | What to try |
|---|---|---|
| `No such file or directory` | Typing error, wrong capital letters or wrong current folder | Add `pwd` and `ls` to the cell to check where you are. In the Terminal tab, use Tab to complete names. |
| `command not found` | Typing error, or the program is not installed | Check the spelling and the capital letters. |
| `Permission denied` | Writing in a protected folder such as `/etc` | Work inside `$LAB_HOME`. Permissions are covered in Lab 3. |
| `rmdir: failed to remove, directory not empty` | The folder still contains files | Delete the files first. Ask your instructor before using `rm -r`. |
| A cell never finishes | The command waits for keys (`nano`, `less`, `top`) | Press the stop button (square) in the notebook toolbar, and use the Terminal tab for this command. |
| The Terminal tab shows `>` and nothing runs | A quote mark was not closed | Press Ctrl + C and type the command again. |
| `tree: command not found` | Lab 0 setup did not finish | Run the Lab 0 setup again, or use `ls -R`. |
